# 1.4 - Clasificación: regresión logística y gradiente descendente

* En este notebook veremos cómo resolver un problema de **clasificación binaria** mediante regresión logística y cómo ajustar sus parámetros con el **optimizador de gradiente descendente**.
* Utilizaremos un conjunto de datos sintético con dos variables de entrada y dos clases parcialmente solapadas.
* En el ejemplo interactivo ejecutaremos el entrenamiento **un epoch cada vez** y observaremos cómo cambia la frontera de decisión.

---
# 1. Clasificación

* Un problema de **clasificación** consiste en predecir una categoría a partir de unas variables de entrada.
* En una clasificación binaria tenemos dos clases, que podemos representar como $y \in \{0,1\}$.
* En nuestro ejemplo utilizaremos dos etiquetas: **clase 0** y **clase 1**.
* Las entradas serán dos características sintéticas, **X₁** y **X₂**. Con dos variables podemos representar las observaciones en un plano y mostrar las regiones asignadas por el modelo.
* Las clases se solapan: algunos puntos de distintas clases se encuentran próximos, por lo que no se pueden separar todos perfectamente con una recta.
* Los puntos representan observaciones sintéticas del conjunto de datos. **No se desplazan ni cambian su clase real**; lo que cambia durante el entrenamiento es la predicción del modelo y, por tanto, la frontera entre las regiones.

---
# 2. Modelo de regresión logística

* Al igual que en la regresión lineal, partimos de una combinación de las variables de entrada y los parámetros:

$$z = \beta_0 + \beta_1 x_1 + \beta_2 x_2$$

* La diferencia es que aplicamos a $z$ la **función sigmoide**, que transforma cualquier valor real en un valor entre 0 y 1:

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

* El **modelo de regresión logística** devuelve la probabilidad estimada de pertenecer a la clase 1:

$$p = f(x_1,x_2)=P(y=1\mid x_1,x_2)=\sigma(\beta_0 + \beta_1 x_1 + \beta_2 x_2)$$

$$P(y=0\mid x_1,x_2)=1-p$$

* Con un **umbral de 0,5**, asignamos la clase 1 si $p\geq 0,5$ y la clase 0 si $p<0,5$. Podemos elegir otro umbral según el problema.
* Para este umbral, la frontera cumple $p=0,5$, equivalente a $z=0$. Con dos variables, esta frontera es una recta.

<div style="background-color:#fff4d6; border-left:5px solid #e6a700; padding:12px 18px; color:#453500;">
<strong>IMPORTANTE: la sigmoide como función de activación*</strong><br>
La suma ponderada de las entradas y el término independiente produce <strong>z</strong>. La sigmoide transforma ese valor en una salida entre 0 y 1. En regresión logística, interpretamos esa salida como la probabilidad estimada de la clase 1 y aplicamos un umbral para decidir la clase.<br>
Una neurona artificial también puede utilizar la <strong>sigmoide como función de activación</strong>; existen otras funciones de activación según la tarea.<br>
<small>* Las funciones de activación están relacionadas con las redes neuronales y las veremos más adelante.</small>
</div>

<div style="background-color:#e8f0fe; border-left:5px solid #2f6bff; padding:12px 18px; color:#1f2a44;">
<strong>IMPORTANTE: parámetros del modelo</strong>
<ul>
<li><strong>β₀:</strong> término independiente.</li>
<li><strong>β₁ y β₂:</strong> coeficientes de las dos variables de entrada.</li>
</ul>
El modelo establece la forma de la predicción; el optimizador ajusta los parámetros.
</div>

---
# 3. Función de pérdida y evaluación

* La **entropía cruzada binaria** mide la discrepancia entre la clase real y la probabilidad estimada por el modelo. Para $N$ observaciones:

$$E(\beta_0,\beta_1,\beta_2)=-\frac{1}{N}\sum_{i=1}^{N}\left[y^{(i)}\log(p^{(i)})+(1-y^{(i)})\log(1-p^{(i)})\right]$$

* Si la clase real es 1, la pérdida disminuye cuando la probabilidad de clase 1 se aproxima a 1. Si la clase real es 0, disminuye cuando esa probabilidad se aproxima a 0.
* El optimizador trata de **reducir la pérdida**. Además, mostraremos la **accuracy**, proporción de puntos cuya clase predicha coincide con la real usando el umbral 0,5.
* Pérdida y accuracy miden cosas diferentes: la primera utiliza probabilidades y la segunda compara clases.

---
# 4. Optimizador: gradiente descendente

* Partimos de unos valores para los parámetros $\beta_0$, $\beta_1$ y $\beta_2$. El optimizador calcula el gradiente de la pérdida y los actualiza para reducirla.
* Si $p^{(i)}$ es la probabilidad estimada por el modelo para la observación $i$, las derivadas son:

$$\frac{\partial E}{\partial \beta_0}=\frac{1}{N}\sum_{i=1}^{N}(p^{(i)}-y^{(i)})$$

$$\frac{\partial E}{\partial \beta_j}=\frac{1}{N}\sum_{i=1}^{N}(p^{(i)}-y^{(i)})x_j^{(i)} \quad (j=1,2)$$

* Actualizamos los parámetros **simultáneamente** a partir de sus valores anteriores:

$$\beta_j^{new}:=\beta_j^{old}-\alpha\frac{\partial E}{\partial\beta_j} \quad (j=0,1,2)$$

<div style="background-color:#fff3cd; border-left:5px solid #e0a800; padding:12px 18px; color:#1f2a44;">
<strong>IMPORTANTE: hiperparámetros del entrenamiento</strong>
<ul>
<li><strong>α (learning rate):</strong> controla el tamaño de cada actualización del optimizador.</li>
<li><strong>Número de epochs:</strong> indica cuántas veces se recorren los datos de entrenamiento.</li>
</ul>
</div>

* En este ejemplo usamos **batch gradient descent**: cada epoch utiliza todas las observaciones y actualiza los parámetros una vez. Una tasa demasiado grande puede hacer que la pérdida aumente.

```text
Inicializar beta_0, beta_1 y beta_2
Para cada epoch:
    Calcular z y la probabilidad con la función sigmoide
    Calcular los errores de todas las observaciones
    Calcular el gradiente y actualizar los parámetros
    Evaluar el modelo actualizado
```

---
# 5. Conjunto de datos sintético

* Generamos **120 observaciones**, 60 de cada clase, con dos variables de entrada. Los dos grupos se solapan para que el problema no sea perfectamente separable.
* Utilizamos una semilla para obtener los mismos datos en cada ejecución y mostramos cinco observaciones.
* Para que el entrenamiento sea más estable, **estandarizamos** ambas variables con la media y la desviación típica de estos datos.
* Los parámetros del modelo se aplican a las **variables estandarizadas**. En la gráfica mostramos las variables originales y transformamos la malla con la misma media y desviación.

````python
from io import BytesIO
import base64
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import ipywidgets as widgets
from IPython.display import display

# Dos grupos parcialmente solapados; la semilla permite repetir el ejemplo.
generador = np.random.default_rng(seed=42)
N_POR_CLASE = 60
clase_0 = generador.normal(loc=[-0.7, -0.5], scale=[1.0, 1.0], size=(N_POR_CLASE, 2))
clase_1 = generador.normal(loc=[0.7, 0.5], scale=[1.0, 1.0], size=(N_POR_CLASE, 2))
X_original = np.vstack([clase_0, clase_1])
y = np.concatenate([np.zeros(N_POR_CLASE, dtype=int), np.ones(N_POR_CLASE, dtype=int)])

# Las transformaciones se calculan con estos datos de entrenamiento.
media = X_original.mean(axis=0)
desviacion = X_original.std(axis=0)
X = (X_original - media) / desviacion

dataset = pd.DataFrame(X_original, columns=['X1', 'X2'])
dataset['Clase'] = y
display(dataset.head())
print(f'Número de observaciones: {len(dataset)}')
````

---
# 6. Implementación de un epoch

* Calculamos el gradiente con todas las observaciones y actualizamos los tres parámetros a la vez.
* La pérdida se calcula directamente a partir de $z$ para mantener la estabilidad numérica. La accuracy utiliza un umbral de 0,5.

````python
def sigmoide(z):
    """Calcula la sigmoide evitando desbordamientos numéricos."""
    z = np.asarray(z, dtype=float)
    resultado = np.empty_like(z)
    positivos = z >= 0
    resultado[positivos] = 1 / (1 + np.exp(-z[positivos]))
    exp_z = np.exp(z[~positivos])
    resultado[~positivos] = exp_z / (1 + exp_z)
    return resultado


def un_epoch(X, y, beta_0, betas, alpha):
    """Actualiza simultáneamente los parámetros con todos los datos."""
    probabilidades = sigmoide(beta_0 + X @ betas)
    errores = probabilidades - y
    gradiente_beta_0 = np.mean(errores)
    gradiente_betas = np.mean(errores[:, np.newaxis] * X, axis=0)
    return beta_0 - alpha * gradiente_beta_0, betas - alpha * gradiente_betas


def calcular_metricas(X, y, beta_0, betas):
    """Devuelve entropía cruzada binaria y accuracy sobre estos datos."""
    z = beta_0 + X @ betas
    perdida = np.mean(np.logaddexp(0, z) - y * z)
    predicciones = (sigmoide(z) >= 0.5).astype(int)
    accuracy = np.mean(predicciones == y)
    return perdida, accuracy
````

---
# 7. Entrenamiento interactivo

* Ejecuta las celdas en orden. En el **epoch 0**, todos los parámetros valen cero y el modelo asigna $p=0,5$ a todos los puntos. Con la regla elegida, predice inicialmente la clase 1 para todos y aún no hay una frontera única.
* Pulsa **«Ejecutar 1 epoch»** para actualizar los parámetros. Puedes cambiar **α** para la siguiente pulsación y usar **«Reiniciar»** para volver al modelo inicial.
* El **fondo** representa la clase predicha por el modelo en cada región. Los puntos conservan el color de su **clase real**: si un punto está sobre una región de otro color, el modelo lo clasifica incorrectamente.
* La recta discontinua representa la frontera $p=0,5$ cuando existe dentro de la gráfica. La segunda gráfica muestra la evolución de la pérdida por epoch. Al comenzar suele disminuir rápidamente y después más despacio, formando un codo.

> La pérdida y la accuracy se calculan sobre los mismos datos usados para entrenar. En un problema real, el preprocesamiento se ajusta solo con entrenamiento y la evaluación final se realiza con datos independientes.

````python
ALPHA_INICIAL = 1.0
estado = {'epoch': 0, 'beta_0': 0.0, 'betas': np.zeros(2), 'epochs': [], 'perdidas': []}

boton_epoch = widgets.Button(description='Ejecutar 1 epoch', button_style='primary',
                             icon='step-forward', layout=widgets.Layout(width='180px'))
boton_reiniciar = widgets.Button(description='Reiniciar', icon='refresh', layout=widgets.Layout(width='130px'))
alpha = widgets.FloatText(value=ALPHA_INICIAL, description='α:', step=0.01,
                          layout=widgets.Layout(width='170px'), style={'description_width': '30px'})
resumen = widgets.HTML()
salida_grafica = widgets.HTML(layout=widgets.Layout(width='100%'))

# Malla en la escala original de las variables sintéticas.
x_min, x_max = X_original[:, 0].min() - 0.5, X_original[:, 0].max() + 0.5
y_min, y_max = X_original[:, 1].min() - 0.5, X_original[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 180), np.linspace(y_min, y_max, 180))
malla_original = np.column_stack([xx.ravel(), yy.ravel()])
malla = (malla_original - media) / desviacion


def registrar_metricas():
    perdida, accuracy = calcular_metricas(X, y, estado['beta_0'], estado['betas'])
    estado['epochs'].append(estado['epoch'])
    estado['perdidas'].append(perdida)
    return perdida, accuracy


def actualizar_vista():
    beta_0, betas = estado['beta_0'], estado['betas']
    perdida, accuracy = calcular_metricas(X, y, beta_0, betas)
    resumen.value = (f"<div style='font-size:19px; line-height:1.8'>"
                     f"<b>Epoch:</b> {estado['epoch']} &nbsp;&nbsp; "
                     f"<b>z:</b> {beta_0:.2f} + {betas[0]:.2f} · x₁ + {betas[1]:.2f} · x₂ &nbsp;&nbsp; "
                     f"<b>Pérdida:</b> {perdida:.4f} &nbsp;&nbsp; "
                     f"<b>Accuracy:</b> {accuracy:.2%} &nbsp;&nbsp; <b>α:</b> {alpha.value:g}</div>")

    z_malla = (beta_0 + malla @ betas).reshape(xx.shape)
    regiones = (sigmoide(z_malla) >= 0.5).astype(int)

    fig, (ax_clases, ax_perdida) = plt.subplots(1, 2, figsize=(12, 5))
    fondo = ListedColormap(['#dbeafe', '#fee2e2'])
    colores = ['#2563eb', '#dc2626']
    ax_clases.pcolormesh(xx, yy, regiones, cmap=fondo, vmin=0, vmax=1, shading='auto')
    if np.any(z_malla < 0) and np.any(z_malla > 0):
        ax_clases.contour(xx, yy, z_malla, levels=[0], colors='#334155', linestyles='--', linewidths=1.5)
    for clase, nombre in [(0, 'Clase real 0'), (1, 'Clase real 1')]:
        puntos = y == clase
        ax_clases.scatter(X_original[puntos, 0], X_original[puntos, 1],
                          c=colores[clase], edgecolors='black', linewidths=0.5,
                          s=35, label=nombre, zorder=3)
    ax_clases.set(xlabel='Variable X₁', ylabel='Variable X₂',
                  xlim=(x_min, x_max), ylim=(y_min, y_max), title='Regiones y clases reales')
    ax_clases.legend(loc='best', fontsize=8)

    ax_perdida.plot(estado['epochs'], estado['perdidas'], 'o-', color='darkorange', markersize=3)
    ax_perdida.set(xlabel='Epoch', ylabel='Entropía cruzada binaria', title='Pérdida por epoch')
    ax_perdida.grid(alpha=0.3)
    fig.tight_layout()
    with BytesIO() as imagen:
        fig.savefig(imagen, format='png', dpi=120, bbox_inches='tight')
        imagen_base64 = base64.b64encode(imagen.getvalue()).decode('ascii')
    plt.close(fig)
    salida_grafica.value = (f'<img src="data:image/png;base64,{imagen_base64}" '
                            'alt="Regiones de clasificación y pérdida" style="width:100%; height:auto; display:block;">')


def ejecutar_epoch(_):
    if not np.isfinite(alpha.value) or alpha.value <= 0:
        resumen.value = '<b style="color:#c00; font-size:19px">Introduce un valor de α mayor que cero y finito.</b>'
        return
    estado['beta_0'], estado['betas'] = un_epoch(X, y, estado['beta_0'], estado['betas'], alpha.value)
    estado['epoch'] += 1
    registrar_metricas()
    actualizar_vista()


def reiniciar(_=None):
    estado.update(epoch=0, beta_0=0.0, betas=np.zeros(2), epochs=[], perdidas=[])
    registrar_metricas()
    actualizar_vista()


boton_epoch.on_click(ejecutar_epoch)
boton_reiniciar.on_click(reiniciar)
alpha.observe(lambda cambio: actualizar_vista(), names='value')
display(widgets.HBox([boton_epoch, boton_reiniciar, alpha]), resumen, salida_grafica)
reiniciar()
````
